```python
from autogen_agentchat.agents import AssistantAgent
from autogen_ext.models.openai import OpenAIChatCompletionClient
import os

model_client = OpenAIChatCompletionClient(
    model="llama-3.1-8b-instant",
    api_key=os.environ["GROQ_API_KEY"],
    base_url="https://api.groq.com/openai/v1",
    model_info={
        "vision": False,
        "function_calling": True,
        "json_output": False,
        "family": "unknown",
        "structured_output": False,
    },
)

writer = AssistantAgent(
    name="Writer",
    model_client=model_client,
    system_message=(
        "You are a professional blog writer. "
        "Write short, clear, engaging blog posts. "
        "When the Critic provides feedback, revise the draft "
        "and produce an improved version."
    ),
)

critic = AssistantAgent(
    name="Critic",
    model_client=model_client,
    system_message=(
        "You are a strict but helpful blog critic. "
        "Review the Writer's draft and give 2-3 specific improvements. "
        "Focus on clarity, structure, examples, grammar, and engagement. "
        "If the writing is already good, reply with APPROVED."
    ),
)

print("Writer and Critic agents created successfully!")
```

```python
import asyncio
from autogen_agentchat.messages import TextMessage

async def run_collaboration():
    # Step 1: Writer creates the first draft
    writer_result = await writer.run(
        task=(
            "Write a 150-word blog post about the benefits of reading. "
            "Make it clear, engaging, and suitable for students."
        )
    )

    draft = writer_result.messages[-1].content

    print("=" * 60)
    print("WRITER - FIRST DRAFT")
    print("=" * 60)
    print(draft)

    # Step 2: Critic reviews the draft
    critic_result = await critic.run(
        task=f"""
Review the following blog post.

BLOG POST:
{draft}

Give 2-3 specific improvements.
If the blog post is already good, reply APPROVED.
"""
    )

    feedback = critic_result.messages[-1].content

    print("\n" + "=" * 60)
    print("CRITIC - FEEDBACK")
    print("=" * 60)
    print(feedback)

    # Step 3: Writer improves the draft
    improved_result = await writer.run(
        task=f"""
Improve the following blog post using the Critic's feedback.

ORIGINAL DRAFT:
{draft}

CRITIC FEEDBACK:
{feedback}

Return only the improved blog post.
Keep it around 150 words.
"""
    )

    improved = improved_result.messages[-1].content

    print("\n" + "=" * 60)
    print("WRITER - IMPROVED VERSION")
    print("=" * 60)
    print(improved)

    # Step 4: Critic checks the improved version
    final_critic_result = await critic.run(
        task=f"""
Review this improved blog post:

{improved}

If it is good enough, reply APPROVED.
Otherwise, give 2-3 specific improvements.
"""
    )

    final_feedback = final_critic_result.messages[-1].content

    print("\n" + "=" * 60)
    print("CRITIC - FINAL REVIEW")
    print("=" * 60)
    print(final_feedback)


await run_collaboration()
```
