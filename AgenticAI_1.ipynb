```python
import requests

def wikipedia_search(query: str) -> str:
    url = "https://en.wikipedia.org/w/api.php"

    params = {
        "action": "query",
        "format": "json",
        "list": "search",
        "srsearch": query,
        "srlimit": 3
    }

    headers = {
        "User-Agent": "Lab1_Agent/1.0 (educational project)"
    }

    response = requests.get(
        url,
        params=params,
        headers=headers,
        timeout=10
    )

    response.raise_for_status()

    data = response.json()

    results = data.get("query", {}).get("search", [])

    if not results:
        return "No Wikipedia results found."

    return "\n\n".join(
        f"{item['title']}: {item['snippet'].replace('<span class=\"searchmatch\">', '').replace('</span>', '')}"
        for item in results
    )

print(wikipedia_search("Albert Einstein theory of relativity"))
```

```python
from langchain_core.tools import Tool

# Working Wikipedia tool
wikipedia_tool = Tool(
    name="Wikipedia",
    func=wikipedia_search,
    description="Look up factual information about people, places, events, and science."
)

# Calculator tool
calculator_tool = Tool(
    name="Calculator",
    func=calculator,
    description="Do arithmetic. Input a maths expression such as 1879 * 2."
)

tools = [wikipedia_tool, calculator_tool]

print("Tools ready:", [tool.name for tool in tools])
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt=(
        "You are a factual assistant. "
        "Use Wikipedia for factual questions. "
        "Use Calculator for arithmetic. "
        "For multi-step questions, use the tools in the correct order."
    )
)

print("Agent created successfully!")
```

```python
question = (
    "Who developed the theory of relativity, "
    "and what is his birth year multiplied by 2?"
)

result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": question
        }
    ]
})

print("FINAL ANSWER:")
print(result["messages"][-1].content)
```

```python
for i, message in enumerate(result["messages"]):
    print(f"\n--- Message {i} ---")
    print("Type:", type(message).__name__)
    print("Content:", message.content)

    if hasattr(message, "tool_calls"):
        print("Tool calls:", message.tool_calls)
```

```python
# STEP 1: Ask Wikipedia
wiki_result = wikipedia_search(
    "Who developed the theory of relativity? What is his birth year?"
)

print("=== WIKIPEDIA ===")
print(wiki_result)

# STEP 2: Extract the known birth year
birth_year = 1879

# STEP 3: Use the Calculator tool
calculation = calculator("1879 * 2")

print("\n=== CALCULATOR ===")
print(calculation)

# STEP 4: Ask the LLM to combine the results
final_response = llm.invoke(
    f"""
Answer the following question using the results from the tools.

Question:
Who developed the theory of relativity, and what is his birth year multiplied by 2?

Wikipedia result:
{wiki_result}

Calculator result:
{calculation}

Give a concise final answer.
"""
)

print("\n=== FINAL ANSWER ===")
print(final_response.content)
```
