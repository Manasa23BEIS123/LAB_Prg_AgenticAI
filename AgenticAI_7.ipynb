```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    api_key="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",
    model="llama-3.1-8b-instant",
    temperature=0
)

print("Groq model connected successfully!")
```

```python
import sqlite3

conn = sqlite3.connect("memory.db")

conn.execute("""
CREATE TABLE IF NOT EXISTS facts (
    key TEXT,
    value TEXT
)
""")

conn.commit()

print("Database ready: memory.db")
```

```python
def remember(key, value):
    conn.execute(
        "INSERT INTO facts (key, value) VALUES (?, ?)",
        (key, value)
    )
    conn.commit()


remember("name", "Riya")
remember("favourite_subject", "Mathematics")

print("Memories saved successfully!")
```

```python
def recall_all():
    rows = conn.execute(
        "SELECT key, value FROM facts"
    ).fetchall()

    return "; ".join(
        f"{key} = {value}" for key, value in rows
    )


memory = recall_all()

print("Recalled memory:", memory)

reply = llm.invoke(
    f"Known facts: {memory}. "
    "Greet the user by name and mention their favourite subject."
)

print("\nAgent response:")
print(reply.content)
```

```python
# Close the old database connection
conn.close()

# Reopen the SQLite database
import sqlite3

new_conn = sqlite3.connect("memory.db")

# Recall memories from the database
rows = new_conn.execute(
    "SELECT key, value FROM facts"
).fetchall()

memory = "; ".join(
    f"{key} = {value}" for key, value in rows
)

print("Memory after new session:", memory)

# Ask the Groq model to use the recalled memory
reply = llm.invoke(
    f"Known facts: {memory}. "
    "Greet the user by name and mention their favourite subject."
)

print("\nAgent response:")
print(reply.content)

new_conn.close()
```
