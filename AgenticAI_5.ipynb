```python
from google.colab import files

uploaded = files.upload()
pdf_name = list(uploaded.keys())[0]

print("Uploaded:", pdf_name)
```

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

pages = PyPDFLoader(pdf_name).load()

splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents(pages)

print("Number of chunks:", len(chunks))
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

db = Chroma.from_documents(
    chunks,
    embeddings
)

print("Vector database ready!")
```

```python
import os
from getpass import getpass
from langchain_groq import ChatGroq

os.environ["GROQ_API_KEY"] = getpass("Enter your Groq API key: ")

llm = ChatGroq(
    api_key=os.environ["GROQ_API_KEY"],
    model="llama-3.1-8b-instant",
    temperature=0
)

print("Groq model connected!")
```

```python
def answer_from_pdf(query):
    docs = db.as_retriever().invoke(query)

    context = "\n".join(d.page_content for d in docs)

    prompt = (
        f"Use ONLY this context to answer.\n"
        f"{context}\n\n"
        f"Question: {query}\n"
        "If the context does not contain the answer, reply exactly: "
        "'I don't know'."
    )

    return llm.invoke(prompt).content
```

```python
def adaptive_answer(question, max_tries=3):
    query = question

    for attempt in range(1, max_tries + 1):
        print(f"Attempt {attempt} with query: {query}")

        answer = answer_from_pdf(query)

        if "i don't know" not in answer.lower():
            return answer

        # Rephrase the query and retry
        query = llm.invoke(
            f"Rephrase this search query differently: {query}"
        ).content

    return "Could not find an answer after several tries."
```

```python
question = "What is the main conclusion of the document?"

answer = adaptive_answer(question)

print("\nFinal Answer:")
print(answer)
```
